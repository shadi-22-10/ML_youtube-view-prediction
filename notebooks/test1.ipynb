Data Pre-Processing

```python
# remove columns which are not necessary for prediction
cols_to_drop = [
    "channel_description",
    "channel_custom_url",
    "channel_published_at",
    "channel_country",
    "channel_have_hidden_subscribers",
    "channel_localized_title",
    "channel_localized_description",
    "video_dimension",
    "video_definition",
    "video_licensed_content",
    "video_duration"
    #remove "video_published_at" if necessary
]

df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])

print("Remaining columns:", df.columns.tolist())
```

```python
# drop rows that contain nan in any column
df_clean = df.dropna()

print("Original shape:", df.shape)
print("After dropna:", df_clean.shape)
```

```python
df_clean.head(10)
```

```python
# show columns, types and a few rows to inspect alignment
print("Columns:", df_clean.columns.tolist())
print("\nDtypes:\n", df_clean.dtypes)
```

```python
numeric_cols = [
    'video_view_count',
    'video_like_count',
    'video_comment_count',
    'channel_view_count',
    'channel_subscriber_count',
    'channel_video_count'
]

# force everything to numeric; strings turn into nan
for col in numeric_cols:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

# drop row where one of these numeric columns is nan
df = df_clean.dropna(subset=numeric_cols)

print("Remaining rows:", df.shape[0])
print("Columns dtypes:\n", df[numeric_cols].dtypes)
```

```python
import re
from collections import Counter
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords")
stop = set(stopwords.words("english"))

def tokenize(text):
    text = re.sub(r"[^a-zA-Z0-9 ]", " ", str(text))
    return [w.lower() for w in text.split() if w.lower() not in stop and len(w) > 2]

all_words = []
df['video_title'].astype(str).apply(lambda x: all_words.extend(tokenize(x)))

word_freq = Counter(all_words).most_common(20)
print(word_freq)

# plot
words, counts = zip(*word_freq)
plt.figure(figsize=(10,5))
sns.barplot(x=list(counts), y=list(words), color="teal")
plt.title("Top 20 Most Common Words in Titles")
plt.xlabel("Frequency")
plt.ylabel("Word")
plt.show()
```
