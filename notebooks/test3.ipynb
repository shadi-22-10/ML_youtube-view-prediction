```python
#Removng columns which are empty which are more than 95% empty
drop_cols = ['description', 'tags', 'liveActualStartTime', 
             'liveScheduledStartTime', 'concurrentViewers']

yt_clean = yt_clean_columns.drop(columns=drop_cols)
yt_clean
```

```python
# Converting columns publistedAt and snapchat_date columns to datetime
yt_clean['publishedAt'] = pd.to_datetime(yt_clean['publishedAt'], errors='coerce')
yt_clean['snapshot_date'] = pd.to_datetime(yt_clean['snapshot_date'], errors='coerce')

yt_clean[['publishedAt', 'snapshot_date']].dtypes
```

```python
# Converting column Duration to seconds
import isodate
import numpy as np

def safe_parse_duration(x):
    if pd.isna(x):
        return np.nan
    try:
        return isodate.parse_duration(x).total_seconds()
    except:
        return np.nan

yt_clean['duration_seconds'] = yt_clean['duration'].apply(safe_parse_duration)

yt_clean[['duration_seconds']].dtypes
```

```python
# Coverting columns caption and licensedContent to boolean datatypes and which are not stored in strings
yt_clean['caption'] = yt_clean['caption'].replace({'false': False, 'true': True}).astype(bool)
yt_clean['licensedContent'] = yt_clean['licensedContent'].replace({'false': False, 'true': True}).astype(bool)

yt_clean[['caption', 'licensedContent']].dtypes
```

```python
# Converting columns channel_name, category, dimension, definition, projection to categorical values
cat_cols = ['channel_name', 'category', 'dimension', 'definition', 'projection'] 
for col in cat_cols: 
    yt_clean[col] = yt_clean[col].astype('category')

yt_clean[['channel_name', 'category', 'dimension', 'definition', 'projection']].dtypes
```

```python
#Feature Engineering
#1. Calculating engagment rate
yt_clean['engagement_rate'] = (
    (yt_clean['likeCount'] + yt_clean['commentCount']) / yt_clean['viewCount']
)
```

```python
#2. Days since publishing
yt_clean['days_since_publish'] = (
    yt_clean['snapshot_date'] - yt_clean['publishedAt']
).dt.days
```

```python
#3. Upload day/month
yt_clean['publish_month'] = yt_clean['publishedAt'].dt.month
yt_clean['publish_day'] = yt_clean['publishedAt'].dt.dayofweek
```

```python
#4. like to view ratio
yt_clean['like_view_ratio'] = yt_clean['likeCount'] / yt_clean['viewCount']
```

```python
#5. comment to view ratio
yt_clean['comment_view_ratio'] = yt_clean['commentCount'] / yt_clean['viewCount']
```

```python
#6. like to comment ratio
yt_clean['like_comment_ratio'] = yt_clean['likeCount'] / (yt_clean['commentCount'] + 1)
```

```python
#7. video age in months
yt_clean['video_age_months'] = yt_clean['days_since_publish'] / 30
```

```python
#8. views per day
yt_clean['views_per_day'] = yt_clean['viewCount'] / (yt_clean['days_since_publish'] + 1)
```

```python
#9. title length
yt_clean['title_length'] = yt_clean['title'].str.len()
```

```python
#10. word count in title
yt_clean['title_length'] = yt_clean['title'].str.len()
```

```python
#11. ? in the title 
# Personal comment - personally noticed videos with open ended titles have more views to capture audience 
# click rate (clickbaiting them into watching the view)

yt_clean['title_length'] = yt_clean['title'].str.len()
```

```python
#12. number in the title 
yt_clean['title_has_number'] = yt_clean['title'].str.contains(r'\d').astype(int)
```

```python
#13. currency symbols in titles
# Again, using numbers for example when the an amount ($Xxxxx) is mentioned or number of days mentioned has a more clickbait-y title. 
# also noticed $ gets more views then another currency symbol used. still not sure why, this is just why scrolling youtube.
import re

currency_pattern = r'[\$\€\£\₹\¥\₩\﷼\₽]'
yt_clean['title_has_currency'] = yt_clean['title'].str.contains(currency_pattern).astype(int)
```

```python
#14. average viewCount by presence of any currency symbol in the title
yt_clean.groupby('title_has_currency')['viewCount'].mean()
```

I wanted to see if mentioning money in video titles would get more people to click. So, I built a tool to scan for currency symbols like $, €, £, and ₹. found that videos with these symbols got significantly more views. On average, they had 3.95 million views, compared to just 1.16 million for videos without any currency symbols. That's a huge 240% increase. tells me that titles about money or finance tend to attract a lot more viewer interest.

```python
#15. avg channel views 
channel_stats = yt_clean.groupby('channel_id')['viewCount'].mean()
yt_clean['channel_avg_views'] = yt_clean['channel_id'].map(channel_stats)
```

```python
#Saving all new features and cleaned dataset to new path
out_path = "cleaned_youtube_dataset.csv"
yt_clean.to_csv(out_path, index=False)
```

```python
#category vs engagement rate vs duration
#this plot will show how engagnment rate varies with different categories and how duration of the videos influiences engagement with each category.

#this suggests that longer videos do not guarantee higher engagement. people might drop off early or the content just might not hold attention. 
#shorter videos seem to land better because they are easier to watch and fit into how people browse.

#overall, engagement connects more to the category and the type of content rather than the raw video length. 
#some categories do well with short formats, while others struggle regardless of duration. 
#it is less about attention span and more about whether the content matches what viewers expect in that category.

plt.figure(figsize=(14, 6))

sns.scatterplot(
    data=yt_clean,
    x='category',
    y='engagement_rate',
    size='duration_seconds',
    sizes=(20, 300),
    alpha=0.5
)

plt.xticks(rotation=45)
plt.title("Engagement Rate vs Category (Bubble size = Duration)")
plt.ylabel("Engagement Rate")
plt.xlabel("Category")
plt.tight_layout()
plt.savefig("plots/engagement_rate_vs_category_vs_duration.png", dpi=300)
plt.show()
```

```python
#enagement rate by category and duration
#by making a group and labelling different durations and shows how engagment looks within each category
#most categories have similar engagement across all duration groups, which backs up the idea that video length is not the main factor driving engagement
#a few categories like DIY and Health show slightly better engagement in short-to-medium videos, while categories like News and Beauty stay low no matter how long the video is

yt_clean['duration_group'] = pd.qcut(
    yt_clean['duration_seconds'],
    q=4,
    labels=["Very Short", "Short", "Long", "Very Long"]
)
```
