```python
#Merge columns in channel_list.csv with youtube_dataset.csv to get all of the required features for the model
import pandas as pd
import numpy as np

import re

#Load youstube_dataset and drop all Unnamed columns
youtube_videos= pd.read_csv("youtube_dataset.csv", low_memory=False)
youtube_videos = youtube_videos.loc[:, ~youtube_videos.columns.str.contains('^Unnamed')]

#Load channel_list.csv 
youtube_channels = pd.read_csv("channel_list.csv", engine = "python", encoding="utf-8", on_bad_lines="skip")

#Extract only the columns needed from channel_list.csv
columns_needed = ["channel_id", "channel_subscriberCount", "channel_viewCount", "channel_videoCount"]

#List comprehension ensures only one row for every channel and drops all duplicates
youtube_channels = youtube_channels[[column for column in columns_needed if column in youtube_channels.columns]].drop_duplicates("channel_id")

df = youtube_videos.merge(youtube_channels, on="channel_id", how="left") #Merge both datasets by channel id
df.head()
```

```python
#Convert all of the required columns into numerical values

df = df.loc[:, ~df.columns.str.contains(r'^Unnamed', case=False, na=False)].copy() #Remove unwanted columns

#Regular expression to detect Youtube API style video durations in the form of a string "PT6M22S" -> 6 minutes, 22 seconds
duration = re.compile(r"^P(?:\d+Y)?(?:\d+M)?(?:\d+D)?(?:T(?:(\d+)H)?(?:(\d+)M)?(?:(\d+)S)?)?$") 

#Function to convert ISO8601 duration strings (format of Youtube API duration) into seconds
def convert_to_seconds(t):
    if not isinstance(t,str):  #If the duration, t is not a string return NaN
        return np.nan 
    minutes=duration.match(t) #Matches t with the regular expression, 'duration' created above
    if not minutes: 
        return np.nan   #If t does not match duration return NaN
    hrs, mins, sec = minutes.groups() #If t matches duration convert the duration string into hours, minutes and seconds
    hrs = int(hrs) if hrs else 0
    mins = int(mins) if mins else 0
    sec = int(sec) if sec else 0
    return 3600*hrs + 60*mins + sec

#Adds a new column for the video length and calls the function to convert the ISO8601 string into seconds 
if "duration" in df.columns: 
    df["video_length"] = df["duration"].apply(convert_to_seconds) 

#If a timestamp exists in the publishedAt column of the df convert the string timestamp into datetime pandas objects
if "publishedAt" in df.columns:
    df["publish_date"] = pd.to_datetime(df["publishedAt"], errors = "coerce") 

#If there is a column containing a snapshot of the date the data was collected convert to datetime format
if "snapshot_date" in df.columns:
    df["snapshot_dt"] = pd.to_datetime(df["snapshot_date"], errors="coerce")
    
    reference_date_time = df["snapshot_dt"] # Use this datetime format as a reference date
else:
    reference_date_time = pd.Timestamp.utcnow() #If there is no snapshot date use today's date

#Subtracts the publish time datetime object from the reference datetime object to obtain a Timedelta object in the form X days HH:MM:SS.mmmmmm    
if "publish_date" in df.columns:
    df["video_age"] = (reference_date_time - df["publish_date"]).dt.days #Extracts the number of days, X from the Timedelta object

#Defines all the columns in the dataframe that should be numeric
columns_to_numeric = ["likeCount", "viewCount", "channel_subscriberCount", "commentCount", "channel_viewCount", 
                      "video_length", "channel_videoCount", "video_age"]

#Converts each of those columns into numerical values
for column in columns_to_numeric:
    if column in df.columns:
        df[column] = pd.to_numeric(df[column], errors = "coerce")

category_columns = [] #Empty list used to store category columns

#If the dataset contains a category column consisting of text convert it into dummy binary columns. 1 if belongs to category and 0 if not
if "category" in df.columns: 
    if df["category"].dtype == "O":
        df=pd.get_dummies(df, columns=["category"],prefix="cat", dummy_na=True)
        category_columns = [column for column in df.columns if column.startswith("cat_")]

    #If category column is already numeric store it and handle invalid values
    else:
        df["category"] = pd.to_numeric(df["category"], errors="coerce")
        
#Feature matrix containing all features that will be used in model training

matrix_features = ["viewCount","likeCount", "video_age", "video_length", "channel_subscriberCount", "channel_viewCount",
                   "commentCount", "channel_videoCount", ]

matrix_features += category_columns #Add category columns to the feature matrix

matrix_X = df[matrix_features].fillna(0) #New dataframe which only contains the columns in feature matrix

print(f"Rows: {len(matrix_X):,}") 
print(f"No of features: {len(matrix_features)}") 
print("Columns: ", matrix_features[:5])

matrix_X.head()
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

drop_features = ["viewCount", "likeCount", "commentCount"]

X = matrix_X.drop(columns = drop_features, errors="ignore")
y = pd.to_numeric(df["viewCount"], errors="coerce")

y_cleaned = y.notna()

X = X.loc[y_cleaned]
y=y.loc[y_cleaned]

log_y = np.log1p(y)

X_train, X_test, y_train, y_test = train_test_split(X, log_y, test_size = 0.2, random_state=42)

rf_model_final = Pipeline([("imp", SimpleImputer(strategy="median")), ("rf", RandomForestRegressor(n_estimators=800, min_samples_split=5,
                                                                                                   min_samples_leaf=2, random_state=42, n_jobs=-1))])

rf_model_final.fit(X_train, y_train)
                                                                        
```

```python
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error

y_predict_log = rf_model_final.predict(X_test)

y_predict_views = np.expm1(y_predict_log)
y_actual_views = np.expm1(y_test)


R2 = r2_score(y_actual_views,y_predict_views)
MAE = mean_absolute_error(y_actual_views,y_predict_views)
RMSE = mean_squared_error(y_actual_views, y_predict_views)

print(f"Mean absolute error: {MAE:,.0f} views")
print(f"R2 score: {R2:,.2f}")
print(f"Root mean squared error: {RMSE:,.0f} views")
```

```python
def SMAPE(y_actual,y_pred):
    SMAPE = np.mean(2*np.abs(y_pred-y_actual)/(np.abs(y_actual) + np.abs(y_pred) + 1e-9)) * 100
    return SMAPE

SMAPE = SMAPE(y_actual_views,y_predict_views)
print(f"SMAPE: {SMAPE:.2f}%")         
```

```python
#Prediction actual vs predicted views alongside input features
prediction_df = pd.DataFrame({"actual_views": y_actual_views, "predicted_views": y_predict_views}) 

prediction_df["percentage_error"] = (((prediction_df["actual_views"] - prediction_df["predicted_views"]).abs())/prediction_df["actual_views"]) * 100

features = matrix_X.loc[X_test.index]

pred_df = features.join(prediction_df)
drop_cols = list(pred_df.filter(like="cat_").columns) + ["viewCount"]

pred_df_final = pred_df.drop(columns = drop_cols)
pred_df_final.head(20)

```
